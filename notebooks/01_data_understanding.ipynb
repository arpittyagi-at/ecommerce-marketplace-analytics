### 7. Understanding the Customers Table

The Customers table contains customer information such as customer IDs, location, city, and state.

The dataset contains both `customer_id` and `customer_unique_id`.

`customer_unique_id` is useful for identifying distinct customers across the marketplace.

```python
print(customers.columns)

len(customers)
print('Rows:',len(customers))
print('unique customer IDs:', customers['customer_id'].nunique())
print('unique "unique customer IDs" :', customers['customer_unique_id'].nunique())
```

### Identifying Repeat Customers

A customer can appear more than once through `customer_unique_id`.

We will count how many times each unique customer appears and identify customers with more than one record.

```python
customer_frequency= customers['customer_unique_id'].value_counts()

repeat_customers= customer_frequency[customer_frequency >1]
print(repeat_customers)
print('Total Repeat Customers :',len(repeat_customers))

print('Repeat Customer %age :' , round(len(repeat_customers)/customers['customer_unique_id'].nunique()*100,2))
```

### 9. Main Table Relationships

The main relationships in the dataset are:

- Customers → Orders using `customer_id`
- Orders → Order Items using `order_id`
- Orders → Payments using `order_id`
- Orders → Reviews using `order_id`
- Order Items → Products using `product_id`
- Order Items → Sellers using `seller_id`
- Products → Category Translation using the product category name

The Orders table is the central table connecting the customer's order with payments, reviews, and purchased items.
